```python
import os
import openai
```

```python
openai.api_key = os.getenv('OPENAI_API_KEY')
```

```python
client = openai.OpenAI()
```

```python
completion = client.chat.completions.create(model = 'gpt-6-luna',
            messages = [{'role': 'system',
                         'content': '''You are Marv, 
                        a chatbot that reluctantly answers questions with 
                        sarcastic responses. '''},
                        {'role': 'user',
                        'content': '''I've recently adopted a dog.
                        Could you suggest some dog names? '''}])
```

```python
print(completion.choices[0].message.content)
```

```python
completion = client.chat.completions.create(model = 'gpt-6-luna',
            messages = [{'role': 'system',
                         'content': '''You are a Marketing Assistant in charge of classifying messages on X as Positive, Negative or Neutral.
                         Reply with "Positive" or "Negative" or "Neutral".
                         If the message cannot be classified, reply with "Error: <reason>".

                        
                        Examples:
                        
                        User: "The Magellan hotel in Tenerife is clean, cosy and ideally located near the beach."
                        Assistant: "Positive".

                        User: "The latest Tarentino's movie is a rotten tomato."
                        Assistant: "Negative"
                        
                        User: "The book is about the cruelty of japanese soldiers during WW2"
                        Assistant: "Neutral" 
                        '''},
                        {'role': 'user',
                        'content': '''The price of the book is 100 USD '''}])
```
