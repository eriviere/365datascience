# Piping Chains and the RunnablePassThrough class

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough
```

```python
chat_template_tools = ChatPromptTemplate.from_template('''What are the five most important tools a {job title} need?
                                                        Answer only by listing the tools.''')

chat_template_strategy = ChatPromptTemplate.from_template('''Considering the tools provided, develop a strategy for effectively mastering them: 
                                                            {tools}''')
```

```python
chat = ChatOpenAI(model='gpt-6-luna', seed=365, temperature=1, max_completion_tokens=10000)
```

```python
string_parser = StrOutputParser()
```

```python
chain_tools = chat_template_tools | chat | string_parser | {'tools': RunnablePassthrough()}
chain_strategy = chat_template_strategy | chat | string_parser
```

## Simple solution: call each chain squentially (copy pasting the first results to the chain_strategy input)

```python
print(chain_tools.invoke({'job title': 'data scientist'}))
```

```python
print(chain_strategy.invoke({'tools': '1. Python\n2. SQL\n3. Jupyter Notebook\n4. Git\n5. Tableau or Power BI'}))
```

## Better solution: combine all chains together

```python
chain_combined = chain_tools | chain_strategy
```

```python
result = chain_combined.invoke({'job title': 'data scientist'})
```

```python
print(result)
```
