# Batching

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
chat_template = ChatPromptTemplate.from_messages([('human',
                                                 '''I've recently adopted a {pet} which is a {breed}, 
                                                 could you suggest several training tips? \n''')])
```

```python
chat_template
```

```python
chat = ChatOpenAI(model='gpt-6-luna', seed=365, temperature=1, max_completion_tokens=1000)
```

```python
chain = chat_template | chat
```

## Single request

```python
chat_result = chain.invoke({'pet': 'dog','breed': 'border collie'})
```

```python
print(chat_result.content)
```

## Batch of several parallelized requests

### First using batch

```python
%%time
chain.batch([{'pet': 'dog','breed': 'border collie'},{'pet': 'dog', 'breed': 'german shepard'}])
```

### Time Comparison using invoke

```python
%%time
chain.invoke({'pet': 'dog','breed': 'border collie'})
```

```python
%%time
chain.invoke({'pet': 'dog', 'breed': 'german shepard'})
```
