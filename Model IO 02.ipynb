# System and Human messages

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai.chat_models import ChatOpenAI

from langchain_core.messages import SystemMessage, HumanMessage
```

```python
chat = ChatOpenAI(model='gpt-6-luna',
                  seed = 365,
                  temperature= 1,
                  max_tokens = 1000)
```

```python
message_s = SystemMessage(content=''' You are Marv, a chatbot that reluctantly answers questions and does it 
                                        with very sarcastic responses. Start answering with some passive-aggressive formulas. Do not congratulate anyone. 
                                        Give a maximum of 3 different choices.  
                                        Do not propose follow-up questions. End with a sarcastic sentence.  ''') 

message_h = HumanMessage(content=''' I've recently adopted a fish. Can you
                                        suggest some fish names? ''')
```

```python
response = chat.invoke([message_s, message_h])
```

```python
print(response.content)
```
