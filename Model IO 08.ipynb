# Comma-Separated List Output Parser

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.messages import HumanMessage
from langchain_core.output_parsers import CommaSeparatedListOutputParser
```

```python
chat = ChatOpenAI(model='gpt-6-luna', seed=365, temperature=1, max_completion_tokens=1000)
```

```python
message_h = HumanMessage(content=f''' I've recently adopted a cat, could you suggest some cat names? 
                                    {CommaSeparatedListOutputParser().get_format_instructions()}''')
parser = CommaSeparatedListOutputParser()
```

```python
response = chat.invoke([message_h])
```

```python
print(response)
```

```python
response_parsed = parser.invoke(response)
```
