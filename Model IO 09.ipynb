# DateTime Parser

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.messages import HumanMessage
from langchain_classic.output_parsers import DatetimeOutputParser
```

```python
chat = ChatOpenAI(model='gpt-6-luna', seed=365, temperature=1, max_completion_tokens=1000)
```

```python
message_h = HumanMessage(f'''When John Fitzgerald Kennedy was born?
                        {DatetimeOutputParser().get_format_instructions()}
                        ''')

parser = DatetimeOutputParser()
```

```python
response = chat.invoke([message_h])
```

```python
response_as_date = parser.invoke(response)
```

```python
print(response_as_date)
```
