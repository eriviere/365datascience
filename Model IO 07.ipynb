# String Output Parser

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai.chat_models import ChatOpenAI
from langchain_core.messages import HumanMessage
from langchain_core.output_parsers import StrOutputParser
```

```python
chat = ChatOpenAI(model= 'gpt-6-luna', seed= 365, temperature= 1, max_completion_tokens= 1000)
```

```python
message_h = HumanMessage(content= ''' Could you tell me an interesting fact I probably didn't know about?  ''')
parser = StrOutputParser()
```

```python
response = chat.invoke([message_h])
```

```python
response_as_text = parser.invoke(response)
```

```python
print(response_as_text)
```
