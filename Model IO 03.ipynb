# AI Messages

```python
%load_ext dotenv
%dotenv
```

```python
from langchain_openai.chat_models import ChatOpenAI

from langchain_core.messages import HumanMessage, AIMessage
```

```python
chat = ChatOpenAI(model='gpt-6-luna',
                  seed = 365,
                  max_tokens = 1000)
```

## Few shots prompting

```python
message_h_dog = HumanMessage(content=''' I've recently adopted a dog. Can you
                                        suggest some dog names? ''')

message_ai_dog = AIMessage(content=''' Well, if we must name the dog: **Milo**, **Pepper**, or **Waffles**. There, the difficult paperwork of naming a dog is practically done. ''')

message_h_cat = HumanMessage(content=''' I've recently adopted a cat. Can you
                                        suggest some cat names? ''')

message_ai_cat = AIMessage(content=''' Well, since naming your cat apparently requires a committee: **Mochi**, **Miso**, or **Chairman Meow**. 
                                        Choose wisely; the cat will ignore you either way. ''')

message_h_fish = HumanMessage(content=''' I've recently adopted a fish. Can you
                                        suggest some fish names? ''')

message_ai_fish = AIMessage(content=''' Well, since the fish isn’t going to name itself, try **Fin Diesel**, **Gillbert**, or **Bubbles**. Truly, the naming process has reached its peak. ''')

message_h_rabbit = AIMessage(content=''' I've recently adopted a mouse. Can you
                                        suggest some mouse names? ''')
```

```python
response = chat.invoke([message_h_dog,
                        message_ai_dog,
                        message_h_cat,
                        message_ai_cat,
                        message_h_fish,
                        message_ai_fish,
                        message_h_rabbit])
```

```python
print(response.content)
```
