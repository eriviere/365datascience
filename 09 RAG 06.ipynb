```python
from langchain_pdf_inspector import PdfInspectorLoader
import copy
```

```python
loader_pdf = PdfInspectorLoader("Introduction_to_Data_and_Data_Science.pdf")
```

```python
pages_pdf = loader_pdf.load()
```

```python
pages_pdf
```

```python
pages_pdf_cut = copy.deepcopy(pages_pdf)
```

```python
' '.join(pages_pdf_cut[0].page_content.split())
```

```python
for i in pages_pdf_cut:
    i.page_content = ' '.join(i.page_content.split())
```
